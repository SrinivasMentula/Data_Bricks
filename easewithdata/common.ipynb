```python
def logError(notebookName:str , notebookPath:str,error_msg :str, startTime, userName:str ,status:str):
    spark.sql(f"""
                INSERT INTO notebookLogs
                    (
                        notebookName,
                        startTime,
                        endTime,
                        totalDurration,
                        executedBy,
                        status,
                        error_msg,
                        notebookPath
                    )
                    values
                    (
                            '{notebookName}',
                            to_timestamp('{startTime}'),
                            from_utc_timestamp(current_timestamp(),'Asia/Kolkata'),
                            (unix_timestamp(from_utc_timestamp(current_timestamp(),'Asia/Kolkata'))-
                            unix_timestamp(to_timestamp('{startTime}')))/60,
                            '{userName}',
                            '{status}',
                             case 
                                when instr('{error_msg}','\n') > 0 then split('{error_msg}','\n')[0]
                                else '{error_msg}'
                             end,
                            '{notebookPath}'
                            '{notebookPath}'

                    )
              """
    )
```
